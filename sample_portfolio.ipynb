```python
import pandas as pd

df = pd.read_csv("model_portfolio.csv")
# Large is > 10 billions
# Medium is between 2 billions and 10 billions.
# Small is < 2 billions.
df['Market Cap Size'] = df['Market Cap'].apply(lambda x: 'Small' if x < 2e9 else ('Medium' if x < 10e9 else 'Large'))
```

```python
df['Market Cap Size'].value_counts()
```

```python
df['Country'].value_counts()
```

```python
df['Sector'].value_counts()
```

```python
df["combined"] = list(zip(df["Market Cap Size"],
                          df["Country"],
                          df["Sector"]))

combined_weight = df['combined'].value_counts(normalize=True)

df['combined_weight'] = df['combined'].apply(lambda x: combined_weight[x])

df_sample = df.sample(800, weights=df['combined_weight'])
```

```python
df_sample['Market Cap Size'].value_counts()
```

```python
df_sample['Country'].value_counts()
```
